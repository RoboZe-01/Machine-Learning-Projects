## Text data processing

```python
import nltk 
from  nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stopwords = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):
    # Handle edge cases
    if pd.isna(text) or not isinstance(text, str):
        return ""
    
    # Lowercase
    text = text.lower()
    
    # Remove punctuation but KEEP SPACES (IMPORTANT!)
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)  # ← Added \s
    
    # Split into words
    words = text.split()
    
    # Remove stopwords
    words = [word for word in words if word not in stopwords]
    
    # Lemmatize
    words = [lemmatizer.lemmatize(word) for word in words]
    
    # Join back
    return " ".join(words)
```

## Text vectorization 

Converting text dato into numerical vectors 

`TfidfVectorizer` is a tool from scikit-learn that converts text data into numerical features that machine learning algorithms can understand. It transforms a collection of text documents into a matrix of TF-IDF features.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df_v1['tags'])
```

```python
tfidf_matrix
```

## Applying cosine similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title,n=10):
    if title not in indices:
        return ['Movie not found']

    idx = indices[title]
    sim_score = cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
    similar_idx = sim_score.argsort()[::-1][1:n+1]
    return df_v1['title'].iloc[similar_idx]
```

```python
recommend('Avatar',5)
```
