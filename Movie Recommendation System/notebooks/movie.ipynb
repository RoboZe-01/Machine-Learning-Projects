## 🥇 Essential (Must-Have Features)
These 6 columns carry 80–90% of the predictive signal for your recommender.

1. **genres**

Why: The single strongest categorical signal. User preferences heavily anchor on genre (e.g., Action vs. Romance).

2. **overview**

Why: Plot summaries contain rich textual information (themes, tropes, settings) ideal for NLP similarity modeling.

3. **title**

Why: Required as the human-readable output label for your recommendations.

4. **id (or imdb_id)**

Why: Essential system primary key for indexing, deduplication, and joining with other datasets.

5. **vote_count**

Why: Crucial quality control. Filters out obscure films with 1 artificial "10/10" rating before calculating scores.

6. **vote_average**

Why: Core metric for quality weighting (e.g., combining with vote_count using IMDb’s weighted rating formula).
